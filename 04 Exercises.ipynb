## Exercise 04.2 (functions and default arguments)

Write a single function named `magnitude` that takes each component of a vector of length 2 or 3 and returns the magnitude.
Use default arguments to handle vectors of length 2 or 3 with the same code. Test your function for correctness against hand calculations for a selection of values.

````python
import math
def magnitude(x,y,z=0.0):
    return (x**2+y**2+z**2)**(1/2)


list_a = [3,4,4,4,3]
list_b = [4,3,3,0.0,4]
list_c = [0.0,0.0,0.0,3.0,4]

for a, b, c in zip(list_a, list_b, list_c):
    mag_2d = magnitude(a, b)
    mag_3d = magnitude(a, b, c)
    print(f"a={a}, b={b}, c={c} -> Mag: {mag_3d:.2f}")
````

````python
## tests ##
assert math.isclose(magnitude(3, 4), 5.0)
assert math.isclose(magnitude(4, 3), 5.0)
assert math.isclose(magnitude(4, 3, 0.0), 5.0)
assert math.isclose(magnitude(4, 0.0, 3.0), 5.0)
assert math.isclose(magnitude(3, 4, 4), 6.403124237)
````

## Exercise 04.3 (functions)

Given the coordinates of the vertices of a triangle, $(x_0, y_0)$, $(x_1, y_1)$ and $(x_2, y_2)$, the area $A$ of the triangle is given by:
$$
A = \left| \frac{x_0(y_1  - y_2) + x_1(y_2 - y_0) + x_2(y_0 - y_1)}{2} \right|
$$
Write a function named `area` that computes the area of a triangle given the coordinates of the vertices.
The order of the function arguments must be (`x0, y0, x1, y1, x2, y2)`.

Test the output of your function against some known solutions.

````python
import math

def area(x0, y0, x1, y1, x2, y2):
    
    # Implement the coordinate geometry area formula
    calculated_area = 0.5 * abs(x0 * (y1 - y2) + x1 * (y2 - y0) + x2 * (y0 - y1))
    return calculated_area

# Tests with known geometrical solutions

# Test 1: Right-angled triangle on the axes with base=3, height=4 
assert math.isclose(area(0, 0, 3, 0, 0, 4), 6.0)

# Test 2: Unit square split diagonally (base=1, height=1) 
assert math.isclose(area(0, 0, 1, 0, 0, 1), 0.5)

# Test 3: Triangle with negative coordinates
# Vertices: (-2, -1), (2, -1), (0, 1) -> base=4, height=2
assert math.isclose(area(-2, -1, 2, -1, 0, 1), 4.0)

# Test 4: Flat line (collinear points)
assert math.isclose(area(0, 0, 1, 1, 2, 2), 0.0)
````

````python
## tests ##
x0, y0 = 0.0, 0.0
x1, y1 = 0.0, 2.0
x2, y2 = 3.0, 0.0
A = area(x0, y0, x1, y1, x2, y2)
assert math.isclose(A, 3.0)
````

## Exercise 04.4 (recursion)

The factorial of a non-negative integer $n$ is expressed recursively by:
$$
n! = 
\begin{cases}
1 & n = 0 \\
(n - 1)! \,n & n > 0
\end{cases}
$$

Develop a function named `factorial` for computing the factorial using recursion.
Test your function against the `math.factorial` function, e.g.

````python
import math
print("Reference factorial:", math.factorial(5))
````

````python
import math

def factorial(n):
   
    # Base case: 0! = 1
    if n == 0:
        return 1
    # Recursive case: n! = n * (n - 1)!
    else:
        return n * factorial(n - 1)


# Test a variety of edge cases and typical values
test_values = [0, 1, 5, 10, 12]

for val in test_values:
    custom_result = factorial(val)
    math_result = math.factorial(val)
    
    # Verify custom function matches Python's built-in maths engine
    assert custom_result == math_result
    print(f"factorial({val:2}) = {custom_result:<8} | math.factorial: {math_result:<8} -> Match!")
# val:2 leaves 2 space gao for that variable and < means l
````

````python
## tests ##
assert factorial(0) == 1
assert factorial(1) == 1
assert factorial(2) == 2
assert factorial(5) == 120

import math
assert factorial(32) == math.factorial(32)
````

### Solution

Define the function for computing $f(x)$:

````python
def my_f(x):
    """Evaluate polynomial function"""
    return x**5 / 10 + x**3 - 10 * x**2 + 4 * x + 7
````

````python
def compute_root(f, x0, x1, tol, max_it):
    for n in range(1, max_it + 1):
        x_mid = (x0 + x1) / 2
        f_mid = f(x_mid)  
        f0 = f(x0)
        if abs(f_mid) < tol or (x1 - x0) / 2 < tol:
            return x_mid, f_mid, n
            
        if f0 * f_mid < 0:
            x1 = x_mid
        else:
            x0 = x_mid
    return x_mid, f_mid, max_it

## tests ##
(x, f, num_it) = compute_root(my_f, x0=0, x1=2, tol=1.0e-6, max_it=1000)

# Test solution for function in Exercise 02
assert math.isclose(x, 1.1568354368209839, abs_tol=1e-6)

# Test for new polynomial
def my_f2(x):
    
    return x**3 - 5*x**2 - 2*x + 10

x2, f2, num_it2 = compute_root(my_f2, x0=4, x1=6, tol=1.0e-6, max_it=1000)
assert math.isclose(x2, 5.0, abs_tol=1e-6)
print(f"Root: {x:.16f} found in {num_it} iterations.")
````

#### Optional extension

Use recursion to write a `compute_root` function that *does not* require a `for` or `while` loop.

````python
import math

def my_f(x):
    return (x**5 / 10 + x**3 - 10 * x**2 + 4 * x + 7)

def compute_root(my_f, x0, x1, tol, max_it, it_count=0):
    
    # 1. Base Case: Increment the iteration count
    it_count += 1
    
    x_mid = (x0 + x1) / 2
    f_mid = my_f(x_mid)
    f0 = my_f(x0)
    
    # 2. Base Case: Stop if tolerance is met or max iterations are reached
    if abs(f_mid) < tol or (x1 - x0) / 2 < tol or it_count >= max_it:
        return x_mid, f_mid, it_count
        
    # 3. Update intervals
    if f0 * f_mid < 0:
        x1 = x_mid
    else:
        x0 = x_mid
    
    # 4. Recursive Step
    return compute_root(my_f, x0, x1, tol, max_it, it_count)

## tests ##
x, f, num_it = compute_root(my_f, x0=0, x1=2, tol=1.0e-6, max_it=1000)

# Test solution for function
assert math.isclose(x, 1.1568354368209839, abs_tol=1e-6)
print(f"Root: {x:.16f} found in {num_it} iterations.")
````
